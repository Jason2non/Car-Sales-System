## 1.5 Loops Deep Dive: `break`, `continue` & List Comprehensions

### `break` — exit a loop early

```python
for number in range(1, 10):
    if number == 5:
        break        # stop the loop entirely
    print(number)
```

### `continue` — skip to the next iteration

```python
for number in range(1, 6):
    if number == 3:
        continue     # skip just this one, keep looping
    print(number)
```

### Nested loops — a loop inside a loop

Useful for grids, tables, and comparing every item to every other item.


````python
for number in range(1, 10):
    if number == 5:
        break        # stop the loop entirely
    print(number)
````

````python
for number in range(1, 10):
    if number == 3:
        continue     # skip just this one, keep looping
    print(number)
````

````python
print("break example:")
for number in range(1, 10):
    if number == 5:
        break
    print(number)

print("\ncontinue example:")
for number in range(1, 6):
    if number == 3:
        continue
    print(number)

print("\nnested loop example (small multiplication table):")
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}")
````

### List comprehensions — building a list in one line

A **list comprehension** is a compact way to build a list from a loop. This:
```python
squares = []
for x in range(5):
    squares.append(x ** 2)
```
becomes:
```python
squares = [x ** 2 for x in range(5)]
```
You can add a condition too:
```python
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
```


````python
squares = [x ** 2 for x in range(5)]
print(squares)

even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print(even_squares)

# Works on any iterable, not just range()
names = ["amina", "brian", "cynthia"]
capitalized = [name.title() for name in names]
print(capitalized)
````

### `lambda` — small, throwaway functions

A `lambda` is a one-line, unnamed function. They're mostly used when a function is needed briefly, like a sorting rule.

```python
square = lambda x: x ** 2
square(5)          # 25

people = [("Amina", 25), ("Brian", 19), ("Cynthia", 32)]
people.sort(key=lambda person: person[1])   # sort by age
```


````python
square = lambda x: x ** 2
print(square(5))

people = [("Amina", 25), ("Brian", 19), ("Cynthia", 32)]
people.sort(key=lambda person: person[1])   # sort by age (2nd item in tuple)
print(people)

# map() and filter() often pair with lambda
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(doubled)
print(evens)
````

## 3.1 Error Handling: `try` / `except` / `finally`

Things go wrong sometimes — a user types text instead of a number, a file doesn't exist, a list index is out of range. Python raises an **exception** when this happens. Without handling, an exception crashes your program.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("You can't divide by zero!")
```

You can catch different exception types separately, and use `finally` for code that should always run (like closing a file):

```python
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("That wasn't a valid number!")
finally:
    print("Done trying.")
```


````python
number = 10/0
````

````python
try:
    number = 10 / 0
except ZeroDivisionError:
    print("You can't divide by zero!")
````

````python
try:
  age = int(input("Enter age"))
  print(age)

except:
    print("Please enter a valid number ")
````

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero.")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

### Catching multiple exception types

```python
try:
    value = int("not a number")
except ValueError:
    print("Invalid number format")
except TypeError:
    print("Wrong type entirely")
```

Common built-in exceptions you'll meet as a beginner:

| Exception | When it happens |
|---|---|
| `ValueError` | Right type, wrong value (e.g. `int("abc")`) |
| `TypeError` | Wrong type used in an operation |
| `ZeroDivisionError` | Dividing by zero |
| `KeyError` | Dictionary key doesn't exist |
| `IndexError` | List index out of range |
| `FileNotFoundError` | File doesn't exist when reading |


````python
def get_dict_value(d, key):
    try:
        return d[key]
    except KeyError:
        print(f"'{key}' not found — returning None instead")
        return None

person = {"name": "Amina", "age": 25}
print(get_dict_value(person, "name"))
print(get_dict_value(person, "email"))
````

## 3.2 Working with Files & JSON

### Reading and writing text files

Python uses `open()` with a **mode**: `"w"` (write, overwrites), `"a"` (append), `"r"` (read). The `with` keyword automatically closes the file for you when done — always prefer it.

```python
with open("notes.txt", "w") as f:
    f.write("Hello from Python!\n")
    f.write("This is line two.\n")

with open("notes.txt", "r") as f:
    contents = f.read()
    print(contents)
```

### JSON — saving structured data (lists, dicts) to a file

Text files are fine for plain text, but real programs usually need to save **structured** data — lists and dictionaries. The `json` module converts Python objects to text (and back) so they can be saved and reloaded exactly as they were.

```python
import json

data = {"name": "Amina", "scores": [85, 90, 78]}

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)     # save

with open("data.json", "r") as f:
    loaded = json.load(f)            # load it back
print(loaded)
```


````python
import json

data = {"name": "Amina", "scores": [85, 90, 78]}

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)

with open("data.json", "r") as f:
    loaded = json.load(f)

print(loaded)
print(type(loaded))   # it's a real Python dict again!
````

## 3.4 Introduction to Object-Oriented Programming: Classes & Objects

Everything you've used so far (`str`, `list`, `dict`) is technically an **object** — a bundle of data plus the functions that work on it. A **class** is a blueprint for creating your own objects.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name        # attribute
        self.breed = breed

    def bark(self):              # method (a function that belongs to the class)
        return f"{self.name} says Woof!"

my_dog = Dog("Rex", "Labrador")   # creating an "instance" of Dog
print(my_dog.name)
print(my_dog.bark())
```

- `class` starts the definition; by convention, class names start with a capital letter.
- `__init__` is a special method that runs automatically when you create a new object — it sets up the initial attributes.
- `self` refers to "this particular object" — every method needs it as the first parameter.


````python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        return f"{self.name} says Woof!"

    def describe(self):
        return f"{self.name} is a {self.breed}"

my_dog = Dog("Rex", "Labrador")
print(my_dog.name)
print(my_dog.bark())
print(my_dog.describe())

# You can create as many instances as you like, each with their own data
another_dog = Dog("Bella", "Poodle")
print(another_dog.describe())
````

## 4.2 A gentle peek at Machine Learning & AI

At its core, machine learning is: **show the computer many examples, let it find the pattern, then predict on new data.** Here's the smallest possible example — predicting an exam score from hours studied.


````python
from sklearn.linear_model import LinearRegression

# Made-up example: hours studied -> exam score
hours_studied = [[1], [2], [3], [4], [5]]
exam_score =     [50,   55,   65,   70,   80]

model = LinearRegression()
model.fit(hours_studied, exam_score)   # "learn" the pattern

predicted_score = model.predict([[6]])  # predict for 6 hours studied
print(f"Predicted score for 6 hours studied: {predicted_score[0]:.1f}")
````

Look closely: this is still just a variable (`model`), a function call (`.fit()`), and a function that returns a value (`.predict()`) — the exact same building blocks from earlier, just wrapped inside a smarter library. That's really all ML is at the code level: functions that learn from data instead of being told exact rules.

## 4.3 Automating repetitive work

A huge share of "automation" is just: **loop + function + something that used to be manual.** For example, imagine generating a personalized report line for every employee automatically instead of typing each one by hand:


````python
def performance_note(row):
    if row["salary"] > 80000:
        return "Senior band"
    elif row["salary"] > 55000:
        return "Mid band"
    else:
        return "Entry band"

for _, row in df.iterrows():
    note = performance_note(row)
    print(f"{row['name']} ({row['department']}): {note}")
````

The mechanics — loop, function, condition — don't change.

# Module 5: Capstone Project — Contact Book Manager
---

Time to build something real. This project uses **every skill from this notebook**:

- Variables, strings, and f-strings
- Functions with parameters and return values
- Lists and dictionaries
- Loops and list comprehensions
- Error handling (`try`/`except`)
- Files and JSON (so your contacts are saved between runs!)
- A `while` loop menu, driven by `input()`

We'll build it in three steps: (1) the core functions, (2) a demo showing them working together, (3) an interactive menu you can run in your own notebook.

### Step 1: The core functions


````python
import json

CONTACTS_FILE = "contacts.json"

def load_contacts():
    """Load contacts from the JSON file. Returns an empty list if the file doesn't exist yet."""
    try:
        with open(CONTACTS_FILE, "r") as f:
            return json.load(f)
    except FileNotFoundError:
        return []

def save_contacts(contacts):
    """Save the contacts list to the JSON file."""
    with open(CONTACTS_FILE, "w") as f:
        json.dump(contacts, f, indent=2)

def add_contact(contacts, name, phone, email):
    """Add a new contact and save immediately."""
    contacts.append({"name": name, "phone": phone, "email": email})
    save_contacts(contacts)
    print(f"Added {name}.")

def list_contacts(contacts):
    """Print every contact, numbered."""
    if not contacts:
        print("No contacts yet.")
        return
    for i, c in enumerate(contacts, start=1):
        print(f"{i}. {c['name']} — {c['phone']} — {c['email']}")

def search_contact(contacts, name):
    """Return every contact whose name contains the search text (case-insensitive)."""
    return [c for c in contacts if name.lower() in c["name"].lower()]

def delete_contact(contacts, name):
    """Remove a contact by exact name match. Returns the updated list and how many were removed."""
    updated = [c for c in contacts if c["name"].lower() != name.lower()]
    removed = len(contacts) - len(updated)
    save_contacts(updated)
    return updated, removed

print("Contact Book functions ready!")
````

### Step 2: A demo — using the functions directly

This shows everything working together without needing to type anything (handy for testing in one go).


````python
contacts = load_contacts()

add_contact(contacts, "Amina Yusuf", "0712345678", "amina@example.com")
add_contact(contacts, "Brian Otieno", "0798765432", "brian@example.com")

print("\n--- All contacts ---")
list_contacts(contacts)

print("\n--- Searching for 'ami' ---")
for c in search_contact(contacts, "ami"):
    print(c)

print("\n--- Deleting Brian Otieno ---")
contacts, removed = delete_contact(contacts, "Brian Otieno")
print(f"Removed {removed} contact(s)")

print("\n--- Contacts after deletion ---")
list_contacts(contacts)
````

### Step 3: The interactive menu

This is the real, usable version — a `while True` loop that keeps showing a menu until you choose to quit. **Run this cell in your own notebook** (it needs keyboard input, so it's set up as a function you call yourself rather than run automatically here).


````python
def menu():
    contacts = load_contacts()
    while True:
        print("\n--- Contact Book ---")
        print("1. Add contact")
        print("2. List contacts")
        print("3. Search contact")
        print("4. Delete contact")
        print("5. Quit")
        choice = input("Choose an option (1-5): ")

        if choice == "1":
            name = input("Name: ")
            phone = input("Phone: ")
            email = input("Email: ")
            add_contact(contacts, name, phone, email)
        elif choice == "2":
            list_contacts(contacts)
        elif choice == "3":
            name = input("Name to search: ")
            results = search_contact(contacts, name)
            if results:
                for c in results:
                    print(c)
            else:
                print("No matches found.")
        elif choice == "4":
            name = input("Name to delete: ")
            contacts, removed = delete_contact(contacts, name)
            print(f"Removed {removed} contact(s).")
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid option, try again.")

# Run the line below in your own notebook to try it interactively:
menu()
print("menu() is defined — uncomment 'menu()' above and run this cell to try it!")
````
